```python
# ANALYSIS - keep only the rows we want, then add up the units
storm = orders[(orders['date'] >= START) & (orders['date'] <= END)]
storm = storm[storm['flavor'] == FLAVOR]
units = storm['units'].sum()
per_store = units / storm['store_id'].nunique()
```

```python
# COLLECTION - read the sales export into a table named orders,
# and set the values this analysis is run with
import pandas as pd

FLAVOR = 'Brown Sugar Cinnamon'
START = '2026-09-06'
END = '2026-09-09'

orders = pd.read_csv('https://blkfin.github.io/it630/fall2026/activities/data/poptart-sales-2026-09-07-partial.csv')
orders.head()  # show the first few rows
```

```python
# REPORTING - the two numbers that go back to Dana
print('storm week units:', units)
print('per store:', round(per_store, 1))
```
